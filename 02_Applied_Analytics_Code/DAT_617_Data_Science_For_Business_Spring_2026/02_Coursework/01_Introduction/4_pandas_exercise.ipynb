## 1) Series from a List
**Question:**
A. Create a Pandas Series from the list `[10, 20, 30, 40, 50]`.
B. Print the Series.

```python
import pandas as pd
# A. Create Series from list
lst = [10, 20, 30, 40, 50]
lst_series = pd.Series(lst)

# B. Print
print (lst_series)
```

## 2) Series from a Dictionary
**Question:**
A. Create a dictionary with student ages: `{"Alice": 20, "Bob": 22, "Charlie": 19}`.
B. Convert it into a Pandas Series.
C. Access the age of `"Bob"`.

```python
# A. Dictionary
dict = {"Alice": 20, "Bob": 22, "Charlie": 19}


# B. Series from dictionary
ages = pd.Series(dict)

# C. Access Bob
print(ages.iloc[1])
print(ages.loc['Charlie'])
```

## 3) Series Methods
**Question:**
Given `scores = pd.Series([90, 85, 88, 92, 76])`:
A. Find the mean.
B. Find the standard deviation.
C. Find the sum.

```python
import pandas as pd
scores = pd.Series([90, 85, 88, 92, 76])
# A. Mean
print(scores.mean())

# B. Standard deviation
print(scores.std())

# C. Sum
print(scores.sum())
```

## 4) Series Attributes
Using the `scores` Series above:
A. Print its `dtype`.
B. Print its `size`.
C. Check if all values are unique.

```python
# A. dtype
print(scores.dtype)

# B. size
print(scores.size)

# C. is_unique
print(scores.is_unique)
```
