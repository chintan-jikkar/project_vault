```python
fortune = pd.read_csv('/Users/chintanjikkar/Desktop/PACE/Spring 26/Data Science for Business/Data/fortune1000_2024.csv', 
                      usecols = ['Rank','Company', 'Ticker', 'Sector', 'Industry','Number_of_employees', 'Revenues_M', 'Profits_M', 'MarketCap_Updated_M', 'Country',
       'HeadquartersCity'] ,index_col='Rank')

fortune['rev_per_emp'] = fortune.Revenues_M/fortune.Number_of_employees
(
fortune.assign(rev_per_emp = fortune.Revenues_M/fortune.Number_of_employees)
    .groupby('Sector')
    .rev_per_emp
    .mean()
)
```

```python
fortune['rev_per_emp'] = fortune.Revenues_M/fortune.Number_of_employees
(
fortune.assign(rev_per_emp = fortune.Revenues_M/fortune.Number_of_employees)
    .groupby('Sector')
    .rev_per_emp
    .aggregate(['mean', 'median', 'std'])
)
```

```python
fortune['rev_per_emp'] = fortune.Revenues_M/fortune.Number_of_employees
(
fortune.assign(rev_per_emp = fortune.Revenues_M/fortune.Number_of_employees)
    .groupby('Sector')
    .rev_per_emp
    .aggregate(['mean', 'median', 'std'])
    .plot(kind = 'bar')
)
```
