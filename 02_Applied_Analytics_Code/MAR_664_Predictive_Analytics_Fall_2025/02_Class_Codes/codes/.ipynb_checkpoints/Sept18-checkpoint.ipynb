```python
team_salary.Win_Percent>50
#we are trying to find those that have win rate greater than 50
#aka boolean mask
```

```python
team_salary.loc[team_salary.Win_Percent>50,:]
#this gives us all the team that have value true for the argument
```

```python
team_salary.loc[team_salary.Win_Percent>50,['Team_Name','Average_Salary']]
#this only gives the specified columns of data for those that are true for the argument provided
```

```python
team_salary.loc[(team_salary.Win_Percent>50) & (team_salary.Average_Salary>3000000),['Team_Name','Average_Salary']]
#tuple coz square brackets doesnt work for multiple arguments(nuance) so you use parenthesis to get it done
```

```python
#we are gonna be optimising line 79
team_salary.Average_Salary.mean()
```

```python
mean_salary = team_salary.Average_Salary.mean()
```

```python
team_salary.loc[(team_salary.Win_Percent>50) & (team_salary.Average_Salary<mean_salary),['Team_Name','Average_Salary']]
#i.e. underpaid but overperformed
```
