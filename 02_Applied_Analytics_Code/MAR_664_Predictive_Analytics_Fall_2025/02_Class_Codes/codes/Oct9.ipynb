```python
import pandas as pd
import numpy as np
true = np.arange(1,11,1)
np.random.seed(1031)
pred = np.arange(1,11,1) + np.random.randint(-1,2,10)
data = pd.DataFrame({'true':true,
              'pred':pred})
data
```

```python
#this is to create a new data stream called e for error and then defining how to get it
(
    data.assign(e = data.true - data.pred)
)
```

```python
#ae is absolute error ie there is no -ves only +ves
(
    data.assign(e = data.true - data.pred,
                ae = np.abs(data.true - data.pred))
)
```

```python
#se for squared error non absolute coz squaring nullifies -ve
(
    data.assign(e = data.true - data.pred,
                ae = np.abs(data.true - data.pred),
               se = (data.true - data.pred)**2)
)
```

```python
#root mean squared error RMSE
(
    data.assign(e = data.true - data.pred,
                ae = np.abs(data.true - data.pred),
               se = (data.true - data.pred)**2)
    .se.mean()
)
```
