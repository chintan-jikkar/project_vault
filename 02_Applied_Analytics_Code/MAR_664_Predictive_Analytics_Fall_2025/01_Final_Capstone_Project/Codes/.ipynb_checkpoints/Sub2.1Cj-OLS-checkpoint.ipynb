```python
#Reading the Data files
import pandas as pd
import numpy as np
import statsmodels.api as sm
import seaborn as sns
import matplotlib.pyplot as plt

analysis = pd.read_csv('/Users/chintanjikkar/Desktop/PACE/Fall 25/Predictive Analytics/Assignments/PAC/Data/analysis_data.csv')
scoring = pd.read_csv('/Users/chintanjikkar/Desktop/PACE/Fall 25/Predictive Analytics/Assignments/PAC/Data/scoring_data.csv')
```

```python
#Importing Libraries and functions to be used 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
#Splitting the analysis data
train = analysis.sample(frac=0.7, random_state=1031)
test = analysis.drop(labels = train.index)
```

```python
train.columns
```

```python
#Figuring out the data types of all variables so that OBJ can be created into dummies
train.dtypes
```

```python
#Specifying the variables in the train data split
y_train = train['monthly_spend']
X_train = train.drop('monthly_spend', axis=1)
```

```python
#Checking for NA in the content
print(X_train.isnull().sum().sort_values(ascending=False).head(10))
```

```python
#Rectifying the NAs
# add missingness indicators (helps models learn if "missing" is informative)
X_train['online_shopping_freq_was_na']  = X_train['online_shopping_freq'].isna().astype(int)
X_train['utility_payment_count_was_na'] = X_train['utility_payment_count'].isna().astype(int)
X_train['education_level_was_na']       = X_train['education_level'].isna().astype(int)

# numeric counts -> median inputes of the rest of data
X_train['online_shopping_freq']  = X_train['online_shopping_freq'].fillna(X_train['online_shopping_freq'].median())
X_train['utility_payment_count'] = X_train['utility_payment_count'].fillna(X_train['utility_payment_count'].median())

# categorical -> mode inputes of the rest of data
edu_mode = X_train['education_level'].mode(dropna=True)
edu_fill = edu_mode.iloc[0] if len(edu_mode) else 'Unknown'
X_train['education_level'] = X_train['education_level'].fillna(edu_fill)
```

```python
#Rechecking for NAs after the recti
print(X_train.isnull().sum().sort_values(ascending=False).head(10))
```

```python
#Segregating the categorical stuff
categorical_features = [
    'gender', 'marital_status', 'education_level',
    'region', 'employment_status', 'card_type'
]
```

```python
#Creating dummies
X_train_dum = pd.get_dummies(
    X_train,
    columns=[c for c in categorical_features if c in X_train.columns],
    drop_first=True,
    prefix_sep='_'
)
```

```python
#Checking on the data
print('Shape before dummies:', X_train.shape)
print('Shape after dummies:',  X_train_dum.shape)
print('Object dtypes remaining (should be 0):', X_train_dum.select_dtypes(include='object').shape[1])
print('Preview columns:', list(X_train_dum.columns[:12]))
print(X_train_dum.dtypes)
```

```python
#Converting bool to int for future use(justincase)
bool_cols = X_train_dum.select_dtypes(include=['bool']).columns
X_train_dum[bool_cols] = X_train_dum[bool_cols].astype(int)

X_ols = sm.add_constant(X_train_dum.drop(columns=['customer_id']))
```

```python
#Using OLS this time around
ols_model = sm.OLS(y_train, X_ols).fit()
print(ols_model.summary())
```

```python
#graph just for the fun of it
sm.graphics.plot_regress_exog(ols_model, 'num_transactions', fig=plt.figure(figsize=(12,8)))
plt.show()
```

```python
#dropping insignificant ones
insignificant = [
    'age',
    'annual_income',
    'online_shopping_freq_was_na',
    'utility_payment_count_was_na',
    'education_level_was_na'
]

X_ols_reduced = sm.add_constant(
    X_train_dum.drop(columns=['customer_id'] + insignificant, errors='ignore')
)
```

```python
#OLS 2nd iteration
ols_reduced = sm.OLS(y_train, X_ols_reduced).fit()
print(ols_reduced.summary())
```

```python
#checking before after insigni dropped
# metrics for before model ( all variables)
rmse_before = np.sqrt(mean_squared_error(y_train, ols_model.fittedvalues))
mae_before  = mean_absolute_error(y_train, ols_model.fittedvalues)
r2_before   = ols_model.rsquared
adjr2_before = ols_model.rsquared_adj

# metrics for after model (w/o insignificant ones)
rmse_after = np.sqrt(mean_squared_error(y_train, ols_reduced.fittedvalues))
mae_after  = mean_absolute_error(y_train, ols_reduced.fittedvalues)
r2_after   = ols_reduced.rsquared
adjr2_after = ols_reduced.rsquared_adj

comparison = pd.DataFrame({
    'Model': ['Before (All Variables)', 'After (W/O Insignificant Ones)'],
    'R2': [round(r2_before, 4), round(r2_after, 4)],
    'Adj R2': [round(adjr2_before, 4), round(adjr2_after, 4)],
    'RMSE': [round(rmse_before, 3), round(rmse_after, 3)],
    'MAE': [round(mae_before, 3), round(mae_after, 3)]
})

print('\nModel Performance Comparison:')
display(comparison)
```

```python
drop_cols = [
    'customer_id','age','annual_income',
    'online_shopping_freq_was_na','utility_payment_count_was_na','education_level_was_na'
]

X_ols_core = sm.add_constant(X_train_dum.drop(columns=drop_cols, errors='ignore'))
ols_core   = sm.OLS(y_train, X_ols_core).fit()

print(ols_core.summary().tables[1])
```

```python
#Checking residuals
# Fitted vs residuals
plt.figure(figsize=(6,4))
sns.scatterplot(x=ols_core.fittedvalues, y=ols_core.resid)
plt.axhline(0, color='red', linestyle='--')
plt.title('Residuals vs Fitted Values')
plt.xlabel('Fitted values')
plt.ylabel('Residuals')
plt.show()

# Histogram of residuals
sns.histplot(ols_core.resid, kde=True)
plt.title('Distribution of Residuals')
plt.show()

# QQ plot
import statsmodels.api as sm
sm.qqplot(ols_core.resid, line='45', fit=True)
plt.show()
```

```python
rmse = np.sqrt(mean_squared_error(y_train, ols_core.fittedvalues))
mae  = mean_absolute_error(y_train, ols_core.fittedvalues)

print(f'OLS | R2={ols_core.rsquared:.4f} AdjR2={ols_core.rsquared_adj:.4f} RMSE={rmse:.3f} MAE={mae:.3f}')
```

```python
#Trimming Outliers using Cooks rule 4SD
# 1) Cook's distance
influence = ols_core.get_influence()
cooks_d = influence.cooks_distance[0]
thr = 4.0 / len(X_ols_core)    # classic threshold

mask_keep = cooks_d < thr
print("Trimmed rows:", int((~mask_keep).sum()), "of", len(mask_keep))

# 2) Refit on trimmed training set
X_trim = X_ols_core[mask_keep]
y_trim = y_train.loc[X_ols_core.index[mask_keep]]

ols_trim = sm.OLS(y_trim, X_trim).fit()

# 3) Metrics comparison (train)
def m(label, model, y, yhat=None):
    if yhat is None: yhat = model.fittedvalues
    rmse = np.sqrt(mean_squared_error(y, yhat))
    mae  = mean_absolute_error(y, yhat)
    print(f"{label} | R2={model.rsquared:.4f} AdjR2={model.rsquared_adj:.4f} RMSE={rmse:.3f} MAE={mae:.3f}")

print("\n--- Comparison ---")
m("Before: OLS core", ols_core, y_train)
m("After : OLS trimmed", ols_trim, y_trim)
```

```python
final_cols = [c for c in X_trim.columns if c != 'const']
X_scoring_final = scoring_dum.reindex(columns=final_cols, fill_value=0)
X_scoring_final = sm.add_constant(X_scoring_final)

# predict with the trimmed model
scoring_pred = ols_trim.predict(X_scoring_final)

# save submission
submission = pd.DataFrame({
    'customer_id': scoring['customer_id'],
    'monthly_spend': scoring_pred
})
submission.to_csv('submission_ols_trimmed_v3.csv', index=False)
print('Saved: submission_ols_trimmed_v3.csv')
```
