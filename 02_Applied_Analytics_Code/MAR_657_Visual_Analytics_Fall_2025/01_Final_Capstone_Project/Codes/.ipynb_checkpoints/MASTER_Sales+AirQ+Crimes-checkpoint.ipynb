```python
sales['borough_merge'] = sales['borough'].astype(str).str.upper()
air['borough_merge'] = air['BORO'].str.upper()
crime['borough_merge'] = crime['BORO_NM'].str.upper()
```

```python
# Borough code mapping (NYC standard)
code_to_borough = {
    '1': 'MANHATTAN',
    '2': 'BRONX',
    '3': 'BROOKLYN',
    '4': 'QUEENS',
    '5': 'STATEN ISLAND'
}

# Map code to name
sales['borough_merge'] = sales['borough_merge'].map(code_to_borough)

# Confirm values
print(sales['borough_merge'].unique())
```

```python
#Recheck
print(sales['borough_merge'].unique())
print(air['borough_merge'].unique())
print(crime['borough_merge'].unique())
```

```python
#Correcting Null situation in CRIME
#null_borough = crime[crime['borough_merge'] == '(NULL)']
#print('Null borough count in crime:', len(null_borough))
#print(null_borough[['Latitude', 'Longitude']].head())
```

```python
# Define approximate borough boundaries (bounding boxes)
borough_bounds = {
    'MANHATTAN':   {'lat_min':40.698, 'lat_max':40.881, 'lon_min':-74.047, 'lon_max':-73.907},
    'BRONX':       {'lat_min':40.785, 'lat_max':40.915, 'lon_min':-73.933, 'lon_max':-73.765},
    'BROOKLYN':    {'lat_min':40.570, 'lat_max':40.739, 'lon_min':-74.041, 'lon_max':-73.833},
    'QUEENS':      {'lat_min':40.541, 'lat_max':40.800, 'lon_min':-73.962, 'lon_max':-73.700},
    'STATEN ISLAND': {'lat_min':40.477, 'lat_max':40.651, 'lon_min':-74.255, 'lon_max':-74.052},
}

def assign_borough(lat, lon):
    for boro, bounds in borough_bounds.items():
        if (
            bounds['lat_min'] <= lat <= bounds['lat_max'] and
            bounds['lon_min'] <= lon <= bounds['lon_max']
        ):
            return boro
    return None  # Not matched

# Only apply to records with null borough
df_null = crime[crime['borough_merge'] == '(NULL)'].copy()
df_null['borough_merge'] = df_null.apply(
    lambda row: assign_borough(row['Latitude'], row['Longitude']), axis=1
)

# Update the original crime DataFrame
crime.update(df_null)
```

```python
print("Null borough count after mapping:", (crime['borough_merge'] == '(NULL)').sum())
print(crime['borough_merge'].unique())
```
