```python
# Drop the column 'Unnamed: 0' from the air_quality dataset
air_quality = air_quality.drop(columns=['Unnamed: 0'])

# Display the first few rows to confirm the column is removed
print(air_quality.head())
```

```python
# Function to clean the 'Time Period' column
def clean_time_period(value):
    if isinstance(value, str):
        # Extract year from labels like 'Annual Average 2005' or 'Summer 2020'
        if 'Annual Average' in value or 'Summer' in value or 'Winter' in value:
            return value.split()[-1]  # Get the year part
        
        # For ranges like '2005-2007', take the first year
        if '-' in value:
            return value.split('-')[0]
        
        # For single years, just return the value
        return value
    return value

# Apply the function to the 'Time Period' column
air_quality['Time Period'] = air_quality['Time Period'].apply(clean_time_period)

# Convert the cleaned 'Time Period' to datetime format
air_quality['Time Period'] = pd.to_datetime(air_quality['Time Period'], format='%Y', errors='coerce')

# Check the result
print(air_quality[['Time Period']].head())
```

```python
air_quality['Year']=air_quality['Time Period'].dt.year
```

```python
grouped=air_quality.groupby(by=['Year','Borough'],as_index=False).agg(
    {'Data Value':'mean'}
)
grouped
```
