```python
import pandas as pd
```

```python
chunk_size = 60000 # API max rows per request
rows = []
offset = 0
while True:
    url = f'https://data.cityofnewyork.us/resource/qb7u-rbmr.csv?$limit={chunk_size}&$offset={offset}'
    print(f'Downloading rows {offset} – {offset+chunk_size}')
    chunk = pd.read_csv(url)
    if chunk.empty:
        break
    rows.append(chunk)
    offset += chunk_size
    if len(chunk) < chunk_size:
        break
complaints = pd.concat(rows, ignore_index=True)
print('Total rows downloaded:', complaints.shape[0])
```
