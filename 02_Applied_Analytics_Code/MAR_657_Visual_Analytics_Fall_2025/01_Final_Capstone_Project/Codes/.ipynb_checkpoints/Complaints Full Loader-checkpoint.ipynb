```python
columns_to_drop = [
    'SUSP_AGE_GROUP', 'SUSP_RACE', 'SUSP_SEX',
    'VIC_AGE_GROUP', 'VIC_RACE', 'VIC_SEX',
    'PARKS_NM', 'HADEVELOPT', 'HOUSING_PSA',
    'Lat_Lon', 'CMPLNT_FR_DT', 'CMPLNT_FR_TM',
    'TRANSIT_DISTRICT', 'STATION_NAME',
    'X_COORD_CD', 'Y_COORD_CD',
    'KY_CD', 'PD_CD', 'PD_DESC',
    'CMPLNT_TO_TM', 'CMPLNT_NUM', 'ADDR_PCT_CD'
    'JURISDICTION_CODE', 'ADDR_PCT_CD', 'JURISDICTION_CODE', 'PATROL_BORO',
    'CMPLNT_TO_DT'
]

complaints = complaints.drop(columns=columns_to_drop, errors='ignore')

print('Columns after dropping:', complaints.columns)
print('Shape after step:', complaints.shape)
```

```python
# Make sure RPT_DT column is string and stripped
complaints["RPT_DT"] = complaints["RPT_DT"].astype(str).str.strip()

# Preview (optional)
print(complaints["RPT_DT"].head(10))

# Extract year (correct format for 12/31/2024 is "%m/%d/%Y")
complaints["YEAR"] = pd.to_datetime(complaints["RPT_DT"], format="%m/%d/%Y", errors="coerce").dt.year

# Check year breakdown
print(complaints["YEAR"].value_counts().sort_index())

# Filter years 2015-2025
complaints = complaints[(complaints["YEAR"] >= 2015) & (complaints["YEAR"] <= 2025)].copy()

# Confirm result
print(complaints.shape)
```
