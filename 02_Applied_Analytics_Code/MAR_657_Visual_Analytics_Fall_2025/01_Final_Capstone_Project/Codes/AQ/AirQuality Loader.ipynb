```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
work_dir = "/Users/chintanjikkar/Desktop/PACE/Fall 25/Visual Analytics/Team Project/Raw Data/Air Quality"
os.chdir(work_dir)
```

```python
df_1980_2005 = pd.read_csv('/Users/chintanjikkar/Desktop/PACE/Fall 25/Visual Analytics/Team Project/Raw data/Air Quality/ad_viz_tile_data1980-2005.csv')
df_2005_2025 = pd.read_csv('/Users/chintanjikkar/Desktop/PACE/Fall 25/Visual Analytics/Team Project/Raw data/Air Quality/ad_viz_tile_data 2005-2025.csv')
df_pm25 = pd.read_csv('/Users/chintanjikkar/Desktop/PACE/Fall 25/Visual Analytics/Team Project/Raw data/Air Quality/ad_viz_tile_data 2005-2025.csv')
df_nta = pd.read_csv('/Users/chintanjikkar/Desktop/PACE/Fall 25/Visual Analytics/Team Project/Raw data/Air Quality/aqe-nta.csv')
```

```python
def clean_columns(df):
    df.columns = df.columns.str.strip().str.replace(' ', '_').str.replace('-', '_')
    return df

df_1980_2005 = clean_columns(df_1980_2005)
df_2005_2025 = clean_columns(df_2005_2025)
df_pm25 = clean_columns(df_pm25)
df_nta = clean_columns(df_nta)
```

```python
for aqi_df in [df_1980_2005, df_2005_2025]:
    aqi_df = aqi_df.replace("Not Available", np.nan)
    aqi_df['AQI_Value'] = pd.to_numeric(aqi_df['AQI_Value'], errors='coerce')
```

```python
aqi_combined = pd.concat([df_1980_2005, df_2005_2025], ignore_index=True)
```

```python
aqi_combined['Date'] = pd.to_datetime(aqi_combined['Date'], errors='coerce', format='%m/%d/%Y')
```

```python
print("aqi_combined shape:", aqi_combined.shape)
print("aqi_combined columns:", aqi_combined.columns)
print("df_pm25 shape:", df_pm25.shape)
print("df_nta shape:", df_nta.shape)
```

```python
print(aqi_combined.head())
print(df_pm25.head())
print(df_nta.head())
```

```python
# Check for missing values in each column of aqi_combined
print(aqi_combined.isna().sum())

# Optionally: Display rows that contain any NA values
print(aqi_combined[aqi_combined.isna().any(axis=1)])
```

```python
# Check which years are present in the merged AQI dataset
aqi_combined['Year'] = aqi_combined['Date'].dt.year
print("Years in aqi_combined:", sorted(aqi_combined['Year'].dropna().unique()))

# For each DataFrame, list columns and show count of missing values per column
for df, name in zip([aqi_combined, df_pm25, df_nta], ['aqi_combined', 'df_pm25', 'df_nta']):
    print(f"\n{name} columns: {df.columns.tolist()}")
    print(f"{name} missing values per column:\n{df.isna().sum()}")

# Optionally, show number of complete cases per file
for df, name in zip([aqi_combined, df_pm25, df_nta], ['aqi_combined', 'df_pm25', 'df_nta']):
    complete_cases = df.dropna().shape[0]
    print(f"{name} complete cases: {complete_cases} out of {df.shape[0]} rows")
```
