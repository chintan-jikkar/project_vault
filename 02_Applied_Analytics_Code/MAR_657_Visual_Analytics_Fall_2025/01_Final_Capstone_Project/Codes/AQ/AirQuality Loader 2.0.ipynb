```python
import pandas as pd
import numpy as np
```

```python
df_1980_2005 = pd.read_csv('/Users/chintanjikkar/Desktop/PACE/Fall 25/Visual Analytics/Team Project/Raw data/Air Quality/ad_viz_tile_data1980-2005.csv')
df_2005_2025 = pd.read_csv('/Users/chintanjikkar/Desktop/PACE/Fall 25/Visual Analytics/Team Project/Raw data/Air Quality/ad_viz_tile_data 2005-2025.csv')
```

```python
for label, df in zip(['1980–2005', '2005–2025'], [df_1980_2005, df_2005_2025]):
    site_ids = df['Site ID'].astype(str).str.strip()
    county_codes = site_ids.str.split('-').str[1]
    print(f"\nCounty code counts for {label}:")
    print(county_codes.value_counts(dropna=False).sort_index())
    print(f"Unique county codes found in {label}:", sorted(county_codes.dropna().unique()))
    missing_county = county_codes.isna().sum()
    if missing_county > 0:
        print(f"Rows with missing county code in {label}: {missing_county}")
        print(df[site_ids.str.split('-').str[1].isna()].head())
```

```python
print("\nOriginal shapes:", df_2005_2025.shape, df_1980_2005.shape)
```

```python
print("Columns 2005–2025:", df_2005_2025.columns.tolist())
print("Columns 1980–2005:", df_1980_2005.columns.tolist())
```

```python
print("Dtypes 2005–2025:\n", df_2005_2025.dtypes)
print("Dtypes 1980–2005:\n", df_1980_2005.dtypes)
```

```python
for df in [df_2005_2025, df_1980_2005]:
    df.columns = [col.strip() for col in df.columns]
    for col in ['Main Pollutant', 'Site Name', 'Site ID', 'Source']:
        df[col] = df[col].astype(str).str.strip()
```

```python
print("\nAfter cleaning column whitespace:")
print("Columns 2005–2025:", df_2005_2025.columns.tolist())
print("Columns 1980–2005:", df_1980_2005.columns.tolist())
```

```python
for aqi_df in [df_2005_2025, df_1980_2005]:
    aqi_df.replace("Not Available", np.nan, inplace=True)
    aqi_df['AQI Value'] = pd.to_numeric(aqi_df['AQI Value'], errors='coerce')
    aqi_df['Date'] = pd.to_datetime(aqi_df['Date'], errors='coerce')
print("\nAfter conversion and NA replacement (non-null counts):")
print(df_2005_2025.count())
print(df_1980_2005.count())
```

```python
borough_map = {
    '005': 'Bronx',
    '047': 'Brooklyn',
    '061': 'Manhattan',
    '081': 'Queens',
    '085': 'Staten Island',
}
for df in [df_2005_2025, df_1980_2005]:
    df['County_Code'] = df['Site ID'].str.split('-').str[1]
    df['Borough'] = df['County_Code'].map(borough_map)
```

```python
print("\nBorough mapping (first 10 rows 2005–2025):")
print(df_2005_2025[['Site ID', 'County_Code', 'Borough']].head(10))
print("Borough value counts (2005–2025):\n", df_2005_2025['Borough'].value_counts(dropna=False))
```

```python
print("\nBorough mapping (first 10 rows 1980–2005):")
print(df_1980_2005[['Site ID', 'County_Code', 'Borough']].head(10))
print("Borough value counts (1980–2005):\n", df_1980_2005['Borough'].value_counts(dropna=False))
```
